```python
# Q1:
'''. Write a loop that makes seven calls to **`print()`**, so we get on the output the following triangle:
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
# # # # # # # #
'''



for i in range(8):
    print("# # # # # # # #")
```

```python
# Q2: Iterate 0 to 10 using for loop,
for i in range(0, 11):
    print(i)

print("-----------")
# do the same using while loop.
i = 0
while i <= 10:
    print(i)
    i += 1

print("-----------")
# Q3: Iterate 10 to 0 using for loop,
for i in range(10, -1, -1):
    print(i)

print("-----------")
# do the same using while loop.
i = 10
while i >= 0:
    print(i)
    i -= 1

print("-----------")
```

```python
'''Q4. Use nested loops to give similar output:

      #
     ###
    #####
   #######
  #########
 ###########
#############
'''

rows = 11

for i in range(rows):
    # print leading spaces
    for j in range(rows - i - 1):
        print(" ", end="")

    # print hashes
    for k in range(2 * i + 1):
        print("#", end="")

    # move to next line
    print()
```

```python
# Q5. Use **`for`** loop to iterate from 0 to 100 and print only even numbers
for i in range(0, 101):
    if i % 2 == 0:
        print(i)
```

```python
# Q6. Use **`for`** loop to iterate from 0 to 100 and print the sum of all numbers.

total = 0

for i in range(0, 101):
    total += i

print("The sum of all numbers is", total)
```

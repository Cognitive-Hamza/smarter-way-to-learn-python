## chp31 Dictionaries: Looping through values

```python
customer_29876 = {"first name": "David", "last name": "Elliott", "address": "4803 Wellesley St."}
```

```python
print(customer_29876["first name"])
print(customer_29876["last name"])
print(customer_29876["address"])
```

```python
for each_value in customer_29876.values():
  print(each_value)
```

## chp32 Dictionaries: Looping through keys

```python
for each_key in customer_29876.keys():
  print(each_key)
```

## chp33 Dictionaries: Looping through key-value pairs

```python
customer_29876 = {
    "first_name": "David",
    "last_name": "Elliot",
    "address": "4803 Wellesley St.",
}
```

```python
for each_key, each_value in customer_29876.items():
  print("The customer's " + each_key + " is " + each_value)
```

## chp 34 Creating a list of dictionaries

```python
customers = [
    {
        "customer_id": 0,
        "first_name": "Jahanzaib",
        "last name": "Ali",
        "address": "Scheme 45",
    },
    {
        "customer_id": 1,
        "first_name": "Omer",
        "last_name": "Khan",
        "address": "Scheme 33",
    },
    {
        "customer_id": 2,
        "first_name": "Wasif",
        "last_name": "Hussain",
        "address": "North Nazimabad",
    }
]
```

```python
print(customers, {1})
```

```python
print(customers[0])   # first customer
print(customers[1])   # second customer
print(customers[2])   # third customer
```

```python
c = customers[1]

print(c["customer_id"])
print(c["first_name"])
print(c["last_name"])
print(c["address"])
```

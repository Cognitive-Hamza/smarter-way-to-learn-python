##chp41 Fuctions

A function is a block of Python code that robotically does the same thing again and again, whenever you invoke its name. It saves you repetitive coding and makes your code easier to understand

```python
# Here the function is named "add_numbers", so whenever we call the function it will then show up/do its functions.
def add_numbers():
  first_number = 29
  second_number = 31
  total = first_number + second_number
  print(total)
add_numbers()
```

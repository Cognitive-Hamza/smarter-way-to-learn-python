##chp37 Creating a dictionary that contains lists

```python
customer_29876 = {
"first name": "David",
"last name": "Elliott",
"address": "4803 Wellesley St.",
"discounts": ["standard", "volume", "loyalty"],
}
```

##chp38 How to get information out of a list within a dictionary

When he(customer_29876) makes a purchase, we want to give him the biggest discount he qualifies for, but only that one discount. If we give him all the discounts he qualifies for, we'll lose money.
These are the discounts:
brother-in-law – 30%
loyalty – 15%
volume – 10%
standard – 5%

```python
if "brother-in-law" in customer_29876["discounts"]:
  discount_amount = .30
elif "loyalty" in customer_29876["discounts"]:
  discount_amount = .15
elif "volume" in customer_29876["discounts"]:
  discount_amount = .10
elif "standard" in customer_29876["discounts"]:
  discount_amount = .05
```

##chp47 Functions: Passing information back from them

You've learned how a function becomes more versatile when you pass information to it so it can deliver a custom job. But a function can do even more. It can pass information back to the calling code.

```python
# Here's a function that calculates and displays sales tax:
calc_tax(sales_total=101.37, tax_rate=.05)
def calc_tax(sales_total, tax_rate):
  tax = sales_total * tax_rate
  print(tax)
```

```python
def calc_tax(sales_total, tax_rate):
  tax = sales_total * tax_rate
  return tax

# Example usage to show a valid output:
print(calc_tax(sales_total=101.37, tax_rate=.86))
```

```python
sales_tax = calc_tax(sales_total=101.37, tax_rate=.05)
print(sales_tax)
```

```python
# The code above can be reduced to:
print(calc_tax(sales_total=104.37, tax_rate=.05))
```
